### A. 💾 FASTQ Download and Extraction

1. Retrieve FASTQ files from GEO

````python
%%bash
while read LINE; do
    # Check if the Fastq file already exists
    echo "Checking if ./Fastq/$LINE exists"
    if [[ ! -f ./Fastq/${LINE}_1.fastq.gz ]] ; then
        echo "$LINE absent"
        # Optional: prefetch to cache SRA files first
        # prefetch $LINE -O ./SRA/ --max-size 80G
        
        parallel-fastq-dump \
          --sra-id $LINE \
          --threads 12 \
          --outdir ./Fastq/ \
          --split-files \
          --gzip
    else
        echo "------> $LINE fastq files are present"    
    fi
done < ./SRR_Acc_List.txt
````

3. 🧬 Read Identity and Cell Ranger Mapping

      According to [10x Genomics SRA guidance](https://kb.10xgenomics.com/hc/en-us/articles/115003802691-How-do-I-prepare-Sequence-Read-Archive-SRA-data-from-NCBI-for-Cell-Ranger), the reads map as follows:

      | File Name               | Read Length | Content                | Role for Cell Ranger |
      | ----------------------- | ----------- | ---------------------- | -------------------- |
      | `SRR8485290_1.fastq.gz` | 8 bp        | Index read (i7)        | ❌ Skip (not used)    |
      | `SRR8485290_2.fastq.gz` | 27 bp       | Barcode + UMI          | ✅ **Read 1 (R1)**    |
      | `SRR8485290_3.fastq.gz` | 99 bp       | cDNA (gene expression) | ✅ **Read 2 (R2)**    |

<br>
<br>

4. 📁 Rename Files for Cell Ranger Compatibility

      Cell Ranger expects filenames in the following format:
      ```
      Convert <sample>_2.fastq.gz	-->      <sample>_S1_L001_R1_001.fastq.gz  # barcode + UMI
      Convert <sample>_3.fastq.gz	-->      <sample>_S1_L001_R2_001.fastq.gz  # cDNA
      ```

````python
%%bash

# Rename files to match the expected format for cellranger
for r1 in ./Fastq/*_2.fastq.gz; do
    srr_id=$(basename "$r1" | cut -d'_' -f1)
    echo $srr_id
    r2="./Fastq/${srr_id}_3.fastq.gz"

    # zcat < "$r1" | head -n 2
    # echo " "
    # zcat < "$r2" | head -n 2

    if [[ -f "$r1" && -f "$r2" ]]; then
        echo "Renaming $srr_id files to match cellranger format"
        mv "$r1" "./Fastq/${srr_id}_S1_L001_R1_001.fastq.gz"
        mv "$r2" "./Fastq/${srr_id}_S1_L001_R2_001.fastq.gz"
    else
        echo "Skipping $srr_id — missing one of the required files"
    fi
done
````

5. 📁 Merge R1 and R2 Files for each Sample

````python
%%bash

# Rename files to match the expected format for cellranger

cd /Volumes/SSD/Datasets/2019_Hosein_KPC/Raw/

# Path to the mapping CSV
CSV="./SraRunTable.csv"
FASTQ_DIR="./Fastq"
MERGED_DIR="./MergedFastq"
mkdir -p "$MERGED_DIR"


# Extract headerless mapping of SRR to GSM
tail -n +2 "$CSV" | cut -d',' -f1,21 | while IFS=',' read -r SRR GSM; do
    echo "$GSM → $SRR"

    # Append R1 and R2 to GSM-level merged files
    cat "${FASTQ_DIR}/${SRR}_S1_L001_R1_001.fastq.gz" >> "${MERGED_DIR}/${GSM}_S1_L001_R1_001.fastq.gz"
    cat "${FASTQ_DIR}/${SRR}_S1_L001_R2_001.fastq.gz" >> "${MERGED_DIR}/${GSM}_S1_L001_R2_001.fastq.gz"
done
````

````python
%%bash

cd "/Volumes/Macintosh HD/Users/oshah/onedrive/Resources/Datasets/2019_Hosein_KPC/Processed/Inhouse/CellRanger/"

# Path to Cell Ranger v6.1.2
CELLRANGER=/home/oshah/tools/cellranger-6.1.2/cellranger

# Path to reference genome
REFERENCE=/home/oshah/tools/cellranger-6.1.2/references/refdata-gex-GRCm39-2024-A/

# Directory containing renamed FASTQ files
FASTQ_DIR=/Volumes/SSD/Datasets/2019_Hosein_KPC/Raw/MergedFastq/

# Run Cell Ranger for each SRR sample
while read GSM_ID; do

  GSM_ID=$(echo "$GSM_ID" | tr -d '\r\n')
  
  if [[ ! -d "${GSM_ID}/outs" ]]; then
    echo "Running Cell Ranger for $GSM_ID"

    ulimit -n 655360

    #  --dry
    $CELLRANGER count \
      --id=${GSM_ID} \
      --transcriptome=$REFERENCE \
      --fastqs=$FASTQ_DIR \
      --sample=$GSM_ID \
      --expect-cells=3000 \
      --localcores=8 \
      --localmem=50
  else
    echo "Skipping $GSM_ID: output already exists"
  fi

done < ./GSM_Acc_List.txt
````
